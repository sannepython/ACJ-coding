```python
import requests

url = "https://en.wikipedia.org/wiki/Business"

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 Chrome/131.0.0.0 Safari/537.36"
}

response = requests.get(
    url,
    headers=headers,
    timeout=30
)

print(response.status_code)
```

```python
html = response.text
print(html[:1000])
```
